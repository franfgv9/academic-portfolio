## Q3. Which countries show a gap between Government Strategy and actual Research and Development capacity?


We want to know whether countries promise more than they achieve, or if their policies are aligned with their actual performance.

We define the strategic gap as:

Strategic Gap = Government Strategy − Average(Research, Development)


```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# === Calcular la brecha estratégica ===
df["Strategy Gap"] = df["Government Strategy"] - df[["Research", "Development"]].mean(axis=1)       # --> Calcula la media fila a fila entre esas dos columnas, es decir, R + D / 2 para cada país y luego calculamos la brecha como GS - media(R+D) para cada fila es decir para cada país del df

# === Seleccionar top 6 países por región según Total score ===
top6_per_region = (
    df.sort_values(["Region", "Total score"], ascending=[True, False])      # Ordena el DataFrame primero por Region (A→Z) (True) y luego por Total score (mayor a menor) (False)
      .groupby("Region")            # lo agrupamos por region
      .head(6)                     # selecciona los primeros 6 de cada grupo
      .copy()                      # se hace una copia para que si se modifica el df original no afecte a este nuevo DataFrame
)

# === Rango REAL de Total score ===
global_min_ts = float(top6_per_region["Total score"].min())
global_max_ts = float(top6_per_region["Total score"].max())

# Escala azul recortada a 45 para que se vea mejor el rango donde está la mayoría de países
scale_min_ts = global_min_ts
scale_max_ts = min(45.0, global_max_ts)     # si algún día todos fueran <=45, no tendría sentido ir más allá

# Escala de azules continua personalizada (no empieza en blanco → los valores bajos se ven mejor)
custom_blues = [
    [0.0,  "#c6dbef"],  # azul muy claro (pero no blanco) --> 0.0: valor mínimo del rango → azul claro
    [0.35, "#9ecae1"],
    [0.7,  "#3182bd"],
    [1.0,  "#08519c"]   # azul muy oscuro --> 1.0: valor máximo del rango → azul muy oscuro
]

# Orden de regiones (se mantiene el mismo orden para los subplots)
regions = ["Middle East", "Americas", "Europe", "Asia-Pacific", "Africa"]

# Layout 2×3 (dejamos el último hueco vacío)
fig = make_subplots(
    rows=2, cols=3,                         # la figura tendrá dos filas y 3 columnas
    subplot_titles=regions,                 # títulos de cada subplot según el orden antes definido
    # Cada subplot tendrá sus propios ejes X e Y, sin compartirlos entre ellos
    shared_xaxes=False,               
    shared_yaxes=False,
    # Controlan el espacio horizontal y vertical entre subplots (valores en proporción del ancho/alto)
    horizontal_spacing=0.16,
    vertical_spacing=0.17
)

# Lista de tuplas (fila, columna) que indican dónde colocar cada región
subplot_pos = [(1,1), (1,2), (1,3), (2,1), (2,2)]

for idx, region in enumerate(regions):                  # recorre cada región con su índice
    group = top6_per_region[top6_per_region["Region"] == region].copy()     # Filtra top6_per_region para quedarte solo con las filas de esa región concreta
    if len(group) == 0:
        continue            # Si no hay datos para esa región, salta a la siguiente iteración (medida de seguridad)
        
    # Ordenar para que el mayor Total score quede arriba  --> no se mantiene el orden del df top_6 ordenado
    group = group.sort_values("Total score", ascending=True)
    # ¡OJO! Al trabajar con barras horizontales (eje y categórico --> empieza de abajo a arriba según el orden definido en la lista), 
    # plotly invierte el eje Y, por ello ordenamos Total score de menor a mayor
    # Ej: eje y = [A, B, C] → se dibuja C arriba, B en medio y A abajo
    
    countries = group["Country"].tolist()               # Convierte la columna Country en una lista de nombres de países → se usará en el eje Y (CATEGÓRICO)
    gaps = group["Strategy Gap"].tolist()
    total_scores = group["Total score"].tolist()
    
    # Valores de color para la escala azul: recortamos a scale_max_ts
    # (cualquier país con TS > 45 se colorea como si tuviera 45 en la escala azul) --> Sin embargo solo EEUU y China sobrepasan ese valor
    color_values = [min(ts, scale_max_ts) for ts in total_scores]
    
    # Índices de países "outliers" con Total score por encima de 45 (EEUU, China, etc.) --> Esta lista se usará luego para dibujar una segunda barra superpuesta en otro color
    outlier_idx = [i for i, ts in enumerate(total_scores) if ts > scale_max_ts]
    
    # Solo mostramos la "palette" (colorbar) en el primer subplot para no duplicarla
    show_colorbar = (idx == 0)

    # Toma la posición del subplot --> row y column
    r, c = subplot_pos[idx]
    
    # --- 1) Trace principal: todos los países con palette azul recortada en [min, 45] ---
    fig.add_trace(go.Bar(               # Añade un trace de tipo barra al subplot en la fila r y columna c
        y=countries,            # eje Y categórico con los nombres de países
        x=gaps,                 # Eje X con valor de la brecha estratégica 
        orientation="h",        # barras horizontales
        marker=dict(
            color=color_values,         # valores recortados para alimentar la escala de color
            colorscale=custom_blues,    # escala de azules personalizada
            cmin=scale_min_ts,          # mínimo real
            cmax=scale_max_ts,          # máximo visualizado (45)
            colorbar=dict(                      # Añade la barra de la palette a la derecha, con título “Total score”
                title="Total score"
            ),
            showscale=show_colorbar     # solo se muestra en el primer subplot
        ),
        name=region,                      
        showlegend=False,                 # No mostramos leyenda, las regiones ya están en los títulos de los subplots
        customdata=total_scores,          # usamos el Total score real en el hover
        hovertemplate="<b>%{y}</b><br>Total Score: %{customdata:.2f}<br>Brecha: %{x:.2f}<extra></extra>"
    ), 
    row=r, 
    col=c)
    
    # --- 2) Trace de outliers: mismas barras pero en rojo ---
    if outlier_idx:
        out_countries = [countries[i] for i in outlier_idx]
        out_gaps = [gaps[i] for i in outlier_idx]
        out_ts = [total_scores[i] for i in outlier_idx]                 # valor real de Total score (sin ser recortado a max 45)

        fig.add_trace(go.Bar(
            y=out_countries,
            x=out_gaps,
            orientation="h",
            marker=dict(
                color="#d62728"    # rojo para destacar (puedes cambiar el color si quieres)
            ),
            showlegend=False,
            customdata=out_ts,
            hovertemplate="<b>%{y}</b><br>Total Score: %{customdata:.2f}<br>Brecha: %{x:.2f}<extra></extra>"
        ),
        row=r,
        col=c)

    # Línea en cero para cada subplot
    fig.add_vline(x=0, line=dict(color="gray", dash="dash"), row=r, col=c)

fig.update_layout(
    height=700,
    # <br><sup>...</sup> para añadir un subtítulo más pequeño en una segunda línea:
    title_text="Gap between Government Strategy and actual R&D capacity<br><sup>Top 6 countries by region according to Total Score - Blue capped at 45, countries > 45 highlighted in red</sup>",
    showlegend=False,                    # Ocultamos la leyenda globalmente
    bargap=0.15,                         # Espacio entre barras dentro de cada subplot (15%)
    barmode="overlay"                    # superponemos la barra roja sobre la azul en los outliers
)

# Título común en el eje x
fig.update_xaxes(title_text="Strategic Gap")

fig.show()
```

CONCLUSION: The analysis shows that several countries have government strategies more ambitious than their actual research and development capacity, especially in the Americas. In contrast, advanced European economies such as the US, Israel, China, South Korea, and the Netherlands maintain more balanced levels between strategy and R&D, suggesting more coherent and mature AI ecosystems. Africa stands out for small gaps, indicating moderate strategies aligned with limited capabilities. Overall, large gaps imply a risk of political expectations not supported by actual scientific capacity.

Q4.1 Load your dataset and create ISO3 codes


```python
import pandas as pd
import pycountry  # pip install pycountry
import numpy as np
import plotly.express as px

# Función auxiliar para mapear nombres de país a código ISO3
def get_iso3(country_name):
    try:
        return pycountry.countries.lookup(country_name).alpha_3         # pycountry.countries.lookup() exige que el nombre esté escrito como aparece en el estándar internacional ISO
    except LookupError:
        return np.nan

df["iso3"] = df["Country"].apply(get_iso3)

# Si algún país queda en NaN, puedes imprimirlos y corregirlos a mano:
print("Países sin ISO3:", df[df["iso3"].isna()]["Country"].unique())
# Luego puedes hacer un diccionario de corrección manual si hace falta
fix = {
    "The Netherlands": "NLD",
    "Russia": "RUS",
    "Turkey": "TUR"
}
df["iso3"] = df.apply(
    lambda row: fix.get(row["Country"], row["iso3"]), axis=1
)
```

Q4.4 Calculate the “Intensity” metric


Q4.4.1 AI per million inhabitants and per trillion GDP


```python
# Evitar divisiones por cero
df = df[(df["GDP_2023"] > 0) & (df["POP_2023"] > 0)].copy()

# AI por millón de habitantes
df["AI_per_million_pop"] = df["Total score"] / (df["POP_2023"] / 1_000_000)

# AI por billón (trillion) de US$ de PIB
df["AI_per_trillion_GDP"] = df["Total score"] / (df["GDP_2023"] / 1_000_000_000_000)
```

Q4.4.2 Normalize each one to a 0–100 scale (min–max)


```python
def minmax_0_100(series):
    return 100 * (series - series.min()) / (series.max() - series.min())

df["AI_pop_norm"] = minmax_0_100(df["AI_per_million_pop"])
df["AI_gdp_norm"] = minmax_0_100(df["AI_per_trillion_GDP"])

# Intensidad final como media de ambas
df["Intensity"] = (df["AI_pop_norm"] + df["AI_gdp_norm"]) / 2
```

We can see that the results of our “Intensity” do not match those in the reference article.

The intensity we have built so far:

* divides or normalizes Total Score by GDP and population
* heavily favors very small countries (Malta, Iceland, Luxembourg…)
* penalizes the giants (USA, China, India)

This is fine if you want to measure “how much AI per person/dollar” there is, but not if our goal is “I want countries similar to Spain in size and economy, but performing better in AI.”


SOLUTION: Now we will compare SPAIN with countries whose GDP and size are between half and double of Spain’s, normalize Intensity so that it is on the same scale as Total Score, and calculate a global metric that gives more weight to Total Score than to Intensity, so that not only small countries benefit, but also those leading the AI market.


```python
import pandas as pd
import numpy as np
import plotly.express as px

# =====================================================
# 1. Función auxiliar de normalización 0–100
# =====================================================
def minmax_0_100(s):
    """Escala una serie numérica al rango [0, 100]."""
    return 100 * (s - s.min()) / (s.max() - s.min())


# =====================================================
# 2. Seleccionar país de referencia (España)
# =====================================================
country_ref = "Spain"

spain_row = df[df["Country"] == country_ref].iloc[0]            # filtra por España y coge la primera fila
spain_gdp   = spain_row["GDP_2023"]
spain_pop   = spain_row["POP_2023"]
spain_total = spain_row["Total score"]

print(f"{country_ref} - GDP_2023: {spain_gdp:,.0f}  |  POP_2023: {spain_pop:,.0f}  |  Total score: {spain_total:.2f}")


# =====================================================
# 3. Definir grupo comparable a España
#    (países con PIB y población entre la mitad y el doble)
#    -> puedes ajustar los factores 0.5 y 2.0 si quieres
# =====================================================
peer = df[
    (df["GDP_2023"].between(spain_gdp * 0.5, spain_gdp * 2.0)) &            # between(a, b) comprueba si el valor de cada país está entre a y b
    (df["POP_2023"].between(spain_pop * 0.5, spain_pop * 2.0))
].copy()

print("Nº de países comparables a España:", len(peer))
print(peer[["Country", "Region", "GDP_2023", "POP_2023"]].head())


# =====================================================
# 4. Normalizar Total score e Intensity DENTRO del grupo comparable
# =====================================================
peer["Total_norm"]     = minmax_0_100(peer["Total score"])
peer["Intensity_norm"] = minmax_0_100(peer["Intensity"])

# Pesos de la métrica global (ajustables)
w_total = 0.7   # peso para Total score (rendimiento absoluto en IA)
w_int   = 0.3   # peso para Intensity (eficiencia relativa)

peer["Global_AI"] = w_total * peer["Total_norm"] + w_int * peer["Intensity_norm"]
# peer["Global_AI"] = w_total * peer["Total score"] + w_int * peer["Intensity_norm"]



# =====================================================
# 5. Países que lo hacen mejor que España en Total score
#    (dentro del grupo comparable) y ordenados por Global_AI
# =====================================================
better_peers = (
    peer[peer["Total score"] > spain_total]
    .sort_values("Global_AI", ascending=False)          # eindexa el DataFrame de 0,1,2,… para tenerlo “limpio”
    .reset_index(drop=True)
)

print("Países comparables a España pero con mejor Total score:")
display(better_peers[["Country", "Region", "Total score", "Intensity", "Global_AI"]])


# =====================================================
# 6. Ranking en barras (Global_AI)
# =====================================================
fig_rank = px.bar(
    better_peers.sort_values("Global_AI"),
    x="Global_AI",
    y="Country",
    orientation="h",
    color="Region",
    hover_data={
        "Total score":":.1f",
        "Intensity":":.1f",
        "GDP_2023":":.2e",
        "POP_2023":":.0f"
    },
    title=f"Countries similar to {country_ref} in size and economy, <br>but with better overall performance in AI",       # <br> salto de linea
    labels={"Global_AI": "Global AI Metrics (70% Total score, 30% Intensity)"}
)
fig_rank.update_layout(template="plotly_white", height=600)
fig_rank.show()

```

To identify countries comparable to Spain but with better AI performance, we first define a reference group consisting of countries with a similar economic and demographic size (GDP and population between half and double that of Spain).
Within this group, we normalize the Total Score and the Intensity metric (AI relative to GDP and population), and construct a weighted global metric:

Global_AI = 0.7 × Total_norm + 0.3 × Intensity_norm

This way, we prioritize absolute AI performance while maintaining a component of relative efficiency. Using this metric, we select countries that outperform Spain in Total Score and rank them by their Global_AI metric, thus obtaining a set of “mirror countries” that can serve as a reference for public policy formulation.


## Q2: What common investment patterns stand out in the four countries with the highest AI performance, and how do they compare with Spain’s current strategy?


```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

pillars = [
    "Talent", "Infrastructure", "Research", "Development",
    "Operating Environment", "Government Strategy", "Commercial"
]

country_focus = "Spain"
countries_to_compare = ["Canada", "South Korea", "Australia", "France"]
subplot_titles = [
    "Spain vs Canada",
    "Spain vs South Korea",
    "Spain vs Australia",
    "España vs France"
]

# Perfil de España
spain_profile = df.loc[df["Country"] == country_focus, pillars].iloc[0].values

# Función auxiliar para obtener el perfil de un país
def get_profile(country):
    return df.loc[df["Country"] == country, pillars].iloc[0].values         # Filtra las filas donde Country es "Spain", selecciona solo las columnas de los pilares, toma la primera fila y devuelve un array numpy con los valores

# ============================
# 2. CREAR FIGURA 2x2 CON POLARES
# ============================
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=subplot_titles,
    specs=[[{"type": "polar"}, {"type": "polar"}],
           [{"type": "polar"}, {"type": "polar"}]],
    horizontal_spacing=0.30,                # Espacio horizontal entre subplots
    vertical_spacing=0.12                   # Espacio vertical entre subplots
)

# Colores
color_spain = "#2ca02c"
other_colors = ["#1f77b4", "#ff7f0e", "#9467bd", "#ffcc00"]

# ============================
# 3. AÑADIR LOS 4 RÁDARES
# ============================
for i, country in enumerate(countries_to_compare, start=1):
    row = (i - 1) // 2 + 1
    col = (i - 1) % 2 + 1

    # Con estas fórmulas, cada país se coloca automáticamente en su sitio:
        # (1,1): Canada
        # (1,2): South Korea
        # (2,1): Australia
        # (2,2): France

    other_profile = get_profile(country)

    # España
    fig.add_trace(
        go.Scatterpolar(
            r=spain_profile,        # Radios --> valores numéricos de España en cada pilar
            theta=pillars,             # Ángulos --> nombre de cada pilar alrededor del círculo
            name="Spain",               # Nombre de la serie --> así aparecerá en la leyenda
            fill="toself",                  # Rellena el área del polígono
            # Ambos usan el mismo verde para España
            line_color=color_spain,
            fillcolor=color_spain,
            opacity=0.6,
            showlegend=(i == 1)  # solo en el primer subplot
             # Esto es un truco:
                # Cuando i == 1 (primer subplot) --> showlegend=True --> España aparece en la leyenda
                # Cuando i > 1 --> False --> España NO vuelve a aparecer en la leyenda
                # Así evito tener la entrada “Spain” repetida 4 veces
        ),
        row=row, col=col                # Indica en qué subplot se dibuja este radar
    )

    # País de comparación
    fig.add_trace(
        go.Scatterpolar(
            r=other_profile,
            theta=pillars,
            name=country,
            fill="toself",
            line_color=other_colors[i-1],       # Color asignado según la posición del país en la lista
            fillcolor=other_colors[i-1],
            opacity=0.6
        ),
        row=row, col=col
    )

    # Configuración de ejes polares (misma escala en todos)
    fig.update_polars(
        row=row, col=col,
        radialaxis=dict(
            range=[0, 100],              # # el radio va de 0 a 100, ponemos marcas (ticks) pero ocultamos las etiqueas numericas --> para no saturar
            tickvals=[0, 20, 40, 60, 80, 100],
            showticklabels=False
        ),
        angularaxis=dict(
            rotation=90,             # Rota el gráfico para que el primer pilar empiece a 90 grados y Dirección “clockwise” → las etiquetas de los pilares van en sentido horario
            direction="clockwise"
        )
    )

# ============================
# 4. ESTILO FINAL
# ============================
fig.update_layout(
    height=900,
    width=1100,
    template="plotly_white",
    legend=dict(
        orientation="v",            # leyenda en pos vertical
        y=0.5,                          # ubicada a la derecha del todo, centrada verticalmente
        x=1.05,
        bgcolor="rgba(255,255,255,0.9)",        # fondo casi blanco con ligera transparencia
        # Borde gris fino
        bordercolor="gray",
        borderwidth=1,
        font_size=11,
        title="Countries"
    ),
    margin=dict(l=150, r=160, t=80, b=50)            # MODIFICAMOS LOS MARGÉNES EN ESTE DISEÑO --> r=160 mayor en la derecha para dejar hueco a la leyenda sin cortar los radares.
)

# Ajustar tamaño y posición de títulos de subplots
for ann in fig.layout.annotations:                  # para cada titulo de subplot
    ann.font.size = 13      # pones el tamaño de letra a 13 (algo más pequeño que el título general de la figura)
    ann.y += 0.03       # mueves el título un poco hacia arriba para que no quede pegado al radar

fig.show()
```

## Q9: In which AI investment pillars does Spain differ from South Korea and Australia? What is the magnitude of this gap?


```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# =====================================================================================
# España vs South Korea y España vs Australia (DOS SUBPLOTS)
# =====================================================================================

# Perfiles numéricos limpios
spain_vals = df[df['Country'] == 'Spain'][pillars].iloc[0].astype(float).values
korea_vals = df[df['Country'] == 'South Korea'][pillars].iloc[0].astype(float).values
aus_vals   = df[df['Country'] == 'Australia'][pillars].iloc[0].astype(float).values

# Diferencias: España - país
diff_korea = spain_vals - korea_vals
diff_aus   = spain_vals - aus_vals

diff_korea_rounded = [round(x, 1) for x in diff_korea]      # Recorre los valores de diff_korea y redondea cada uno a 1 decimal
diff_aus_rounded   = [round(x, 1) for x in diff_aus]

# Colores: rojo si España está por debajo, verde si por encima
colors_korea = ['crimson'  if x < 0 else 'darkgreen' for x in diff_korea]
colors_aus   = ['crimson'  if x < 0 else 'darkgreen' for x in diff_aus]

# FIGURA CON 2 SUBPLOTS HORIZONTALES
fig8c = make_subplots(
    rows=1, cols=2,
    shared_yaxes=True,      # Comparte el eje Y entre los dos subplots --> los pilares (eje Y) serán los mismos y estarán alineados
    subplot_titles=("Spain vs South Korea", "Spain vs Australia")
)

# --- Subplot 1: España vs South Korea ---
fig8c.add_trace(
    go.Bar(
        y=pillars,
        x=diff_korea,
        orientation='h',
        marker_color=colors_korea,      # para cada barra, color según si España está por encima o por debajo
        text=diff_korea_rounded,        # Texto que aparece sobre cada barra (la diferencia numérica redondeada)
        textposition='outside',            # Sitúa el texto fuera del extremo de la barra (a la derecha si es positiva, a la izquierda si es negativa)
        textfont=dict(size=13, color="black"),
        name='Spain - South Korea', 
        showlegend=False                # Porque no queremos tener una leyenda por subplot sino una global (por eso hacemos acontinuacnion dos trazas mas)
    ),
    row=1, col=1            # traza numero 1 en fila 1, columna 1
)

# Línea x=0 en el primer subplot
fig8c.add_vline(x=0, line_width=2, line_color="black", row=1, col=1)

# --- Subplot 2: España vs Australia ---
fig8c.add_trace(
    go.Bar(
        y=pillars,
        x=diff_aus,
        orientation='h',
        marker_color=colors_aus,
        text=diff_aus_rounded,
        textposition='outside',
        textfont=dict(size=13, color="black"),
        name='Spain - Australia', 
        showlegend=False 
    ),
    row=1, col=2                # traza numero 2 en fila 1, columna 2
)

# === SOLO PARA LA LEYENDA ===
fig8c.add_trace(
    go.Bar(
        x=[0], y=[""],
        marker_color='darkgreen',           # para que salga el cuadradito verde en la leyenda
        name="Spain invests more",
        visible='legendonly'      # ¡OJO! “no dibujes la barra en el gráfico, muéstrala solo en la leyenda”
    ),
    row=1, col=1        # asociado al subplot 1
)

fig8c.add_trace(
    go.Bar(
        x=[0], y=[""],
        marker_color='crimson',
        name="Spain invests less",
        visible='legendonly'
    ),
    row=1, col=1
)

# Línea x=0 en el segundo subplot
fig8c.add_vline(x=0, line_width=2, line_color="black", row=1, col=2)

# Ejes y estética
fig8c.update_xaxes(
    title="Difference",
    gridcolor="lightgray",
    row=1, col=1
)
fig8c.update_xaxes(
    title="Difference",
    gridcolor="lightgray",
    row=1, col=2
)

fig8c.update_yaxes(
    autorange="reversed",  # para que el primer pilar aparezca arriba --> Invierte el orden de las categorías en el eje Y
    tickfont=dict(size=12),
    row=1, col=1
)

fig8c.update_layout(
    height=550,
    width=1100,
    template="plotly_white",
    font=dict(family="Arial", size=14, color = "Black"),
    margin=dict(l=140, r=60, t=90, b=60),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=-0.25,
        xanchor="center",
        x=0.5,
        title=dict(text="Comparation")
    ),
    legend_font_size=11,        # tamaño del texto de la leyenda
    legend_title_font_size=12,              # tamaño del texto "Comparación"
    title=dict(
        text="COMPARATION FOR PILARS",
        x=0.5,
        y=0.95   # (0=abajo, 1=arriba)
    )
)

fig8c.show()
```
